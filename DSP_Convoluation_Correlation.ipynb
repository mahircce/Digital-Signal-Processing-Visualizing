```python
import numpy as np
import matplotlib.pyplot as plt

# Define the signals
x = np.array([1, -2, 3, -4, 3, 2, 1])
h = np.array([3, 2, 1, -2, 1, 0, -4, 0, 3])

# Compute discrete convolution
y = np.convolve(x, h)

# Calculate time indices (assuming start at index 0)
n_x = np.arange(len(x))
n_h = np.arange(len(h))
n_y = np.arange(len(y))  # Length is len(x) + len(h) - 1

print(f"Convolution result: {y}")
```

```python
# Plotting the signals
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 12))

# Plot x[n]
ax1.stem(n_x, x)
ax1.set_title('Signal x[n]')
ax1.set_xlabel('n')
ax1.set_ylabel('Amplitude')
ax1.grid(True)

# Plot h[n]
ax2.stem(n_h, h)
ax2.set_title('Signal h[n] (Impulse Response)')
ax2.set_xlabel('n')
ax2.set_ylabel('Amplitude')
ax2.grid(True)

# Plot y[n]
ax3.stem(n_y, y)
ax3.set_title('Convolution Result y[n] = x[n] * h[n]')
ax3.set_xlabel('n')
ax3.set_ylabel('Amplitude')
ax3.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
N = 10
f = 0.1
n = np.arange(N)

# Generate signals
x1 = np.sin(2 * np.pi * f * n + 0)
x2 = np.sin(2 * np.pi * f * n + np.pi / 4)

# Compute cross-correlation
# 'full' mode returns a correlation of length 2*N - 1
corr = np.correlate(x1, x2, mode='full')

# Calculate lag indices
lags = np.arange(-N + 1, N)

print(f"Cross-correlation result: {corr}")
```

```python
# Plotting
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 12))

# Plot x1
ax1.stem(n, x1)
ax1.set_title('Sinusoid 1 (Phase = 0)')
ax1.set_xlabel('n')
ax1.set_ylabel('Amplitude')
ax1.grid(True)

# Plot x2
ax2.stem(n, x2)
ax2.set_title('Sinusoid 2 (Phase = pi/4)')
ax2.set_xlabel('n')
ax2.set_ylabel('Amplitude')
ax2.grid(True)

# Plot correlation
ax3.stem(lags, corr)
ax3.set_title('Cross-correlation of Sinusoid 1 and 2')
ax3.set_xlabel('Lag')
ax3.set_ylabel('Correlation')
ax3.grid(True)

plt.tight_layout()
plt.show()
```
