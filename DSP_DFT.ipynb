```python
import numpy as np

# Define the sequence
x = np.array([1, 0, 0, 1])

# Compute the 4-point DFT
X = np.fft.fft(x, n=4)

# Print the input and output
print("Input sequence x(n):", x)
print("4-point DFT X(k):", X)

# Display magnitude and phase for clarity
for k, val in enumerate(X):
    print(f"X({k}) = {val.real:.4f} + {val.imag:.4f}j  (Magnitude: {np.abs(val):.4f}, Phase: {np.angle(val):.4f} rad)")
```

```python
import numpy as np

def compute_dft(x):
    N = len(x)
    X = np.zeros(N, dtype=complex)
    for k in range(N):
        for n in range(N):
            X[k] += x[n] * np.exp(-2j * np.pi * k * n / N)
    return X

# Define the sequence
x = np.array([1, 0, 0, 1])

# Calculate DFT using the custom function
X_custom = compute_dft(x)

# Print the results
print("Input sequence x(n):", x)
print("4-point DFT (Custom implementation):")
for k, val in enumerate(X_custom):
    # Using np.round to clean up small floating-point precision noise (e.g., -0.0000j)
    real_part = np.round(val.real, 10)
    imag_part = np.round(val.imag, 10)
    print(f"X({k}) = {real_part} + {imag_part}j")
```
