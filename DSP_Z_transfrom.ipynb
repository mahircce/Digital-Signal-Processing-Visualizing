```python
# prompt: determine Z transform of a discrete signal

import sympy

# Define the symbolic variable for the Z-transform
z = sympy.symbols('z')

# Define the discrete signal (example: x[n] = a^n u[n])
n = sympy.symbols('n', integer=True)
a = sympy.symbols('a')
x_n = a**n

# Compute the Z-transform using sympy's summation function
X_z = sympy.Sum(x_n * (1/z)**n, (n, 0, sympy.oo))

# Simplify the expression
X_z = X_z.doit()

# Print the Z-transform
print("Z-transform of x[n] =", X_z)

# Example usage with specific values
a_value = 0.5
X_z_specific = X_z.subs({a: a_value})

print(f"Z-transform of x[n] with a = {a_value}:", X_z_specific)
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Assuming x[n] = a^n for n >= 0 (you can change this)
a = 0.5  # Example value for 'a'
n_values = np.arange(0, 10)  # Values of 'n' to plot
x_n_values = a**n_values

# Plot x[n]
plt.figure(figsize=(10, 4))
# Removed 'use_line_collection=True' for compatibility with older Matplotlib versions
plt.stem(n_values, x_n_values)
plt.xlabel('n')
plt.ylabel('x[n]')
plt.title('Discrete Signal x[n]')
plt.grid(True)
plt.show()


# Plot the magnitude and phase of the Z-transform (using the example from the previous code)

# Define the frequency range for the plot
w = np.linspace(-np.pi, np.pi, 500)

# Calculate the magnitude and phase of X(z) on the unit circle
X_z_magnitude = 1/(1-a*np.exp(-1j * w))
X_z_phase = np.angle(X_z_magnitude)

# Create subplots for the magnitude and phase
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(w, abs(X_z_magnitude))
plt.xlabel('ω (radians)')
plt.ylabel('|X(e^jω)|')
plt.title('Magnitude of Z-transform')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(w, X_z_phase)
plt.xlabel('ω (radians)')
plt.ylabel('∠X(e^jω)')
plt.title('Phase of Z-transform')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# prompt: draw pole zero of above z transform

import sympy
import matplotlib.pyplot as plt
import numpy as np

# Define the symbolic variable for the Z-transform
z = sympy.symbols('z')

# Define the discrete signal (example: x[n] = a^n u[n])
n = sympy.symbols('n', integer=True)
a = sympy.symbols('a')
x_n = a**n

# Compute the Z-transform using sympy's summation function
X_z = sympy.Sum(x_n * (1/z)**n, (n, 0, sympy.oo))

# Simplify the expression
X_z = X_z.doit()

# Print the Z-transform
print("Z-transform of x[n] =", X_z)

# Example usage with specific values
a_value = 0.5
X_z_specific = X_z.subs({a: a_value})

print(f"Z-transform of x[n] with a = {a_value}:", X_z_specific)

# --- Pole-Zero Plot ---
# The Z-transform X_z_specific is 1 / (1 - a*z^(-1))  or z / (z - a)
# Pole: z = a
# Zero: z = 0

pole = a_value
zero = 0

plt.figure(figsize=(6, 6))
plt.scatter(np.real(pole), np.imag(pole), marker='x', s=100, label='Pole')
plt.scatter(np.real(zero), np.imag(zero), marker='o', s=100, label='Zero')

# Draw the unit circle
unit_circle = plt.Circle((0, 0), 1, color='gray', fill=False, linestyle='--')
plt.gca().add_patch(unit_circle)

plt.xlabel('Real')
plt.ylabel('Imaginary')
plt.title('Pole-Zero Plot')
plt.grid(True)
plt.axis('equal')
plt.legend()
plt.xlim([-1, 1])  # Adjust limits if needed
plt.ylim([-1, 1])  # Adjust limits if needed
plt.show()
```
